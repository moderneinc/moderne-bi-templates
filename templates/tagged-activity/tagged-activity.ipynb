# Tagged Activity

Committed recipe output grouped by a trace tag — the example uses `tag.changeSetId`, but any `tag.<key>` column works.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load sample data — replace with your own query results
df = pd.read_csv("../../samples/tagged-activity.csv")
df = df.sort_values("successful_commits", ascending=False)
```

```python
fig, ax1 = plt.subplots(figsize=(14, 7))

x = np.arange(len(df))
bar_width = 0.4

# Grouped bars for commit and repo counts per change set
ax1.bar(x - bar_width / 2, df["successful_commits"], width=bar_width,
        color="#283AF7", alpha=0.8, label="Successful Commits")
ax1.bar(x + bar_width / 2, df["repos_changed"], width=bar_width,
        color="#E93D82", alpha=0.8, label="Repos Changed")

ax1.set_xlabel("Change Set", color="black")
ax1.set_ylabel("Count", color="black")
ax1.set_xticks(x)
ax1.set_xticklabels(df["change_set"], rotation=45, ha="right")
ax1.tick_params(axis="both", labelcolor="black")

# Secondary y-axis for estimated hours saved
ax2 = ax1.twinx()
ax2.plot(x, df["estimated_hours_saved"], color="#7C3AED", marker="D",
         linewidth=2, linestyle="-", label="Est. Hours Saved")
ax2.set_ylabel("Estimated Hours Saved", color="black")
ax2.tick_params(axis="y", labelcolor="black")

ax1.set_title("Tagged Activity (by change set)", color="black")

# Combine legends from both axes
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper right", fontsize=8)

plt.tight_layout()
plt.show()
```
